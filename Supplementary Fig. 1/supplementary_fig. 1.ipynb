# Supplementary Fig. 1: model training loss

## Purpose

Display pre-training and supervised fine-tuning loss trajectories for Phyto-Chatbot and Phyto-Reasoner.

## Inputs and provenance

The four tracked JSON files in `Model-Loss-for_plot/` contain the training-log snapshots. The notebook plots their existing step/loss values; it does not retrain a model or replace training loss with evaluation loss.

## Run and outputs

Run all cells in this directory with the project Python kernel and the pinned Plotly/Kaleido environment. Set `PHYTOMNI_SAVE=1` to save figures; the default run displays plots without writing figures. Four PDF/PNG pairs are written under `output/supplementary_fig.1a` through `output/supplementary_fig.1d`, using the descriptive loss basenames below.

```python
import os
SAVE_FIGS = os.getenv("PHYTOMNI_SAVE") == "1"
if SAVE_FIGS:
    os.makedirs("output", exist_ok=True)
```

```python
from json import load

import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio
import plotly.offline as offline
```

```python
offline.init_notebook_mode(connected=True)
pio.kaleido.scope.default_format = 'pdf'
pio.kaleido.scope.mathjax = None
```

```python
with open('Model-Loss-for_plot/Phyto-Chatbot-Pretrain.loss.json', 'r') as open_json:
    chat_pretrain_loss = load(open_json)
chat_pretrain_loss_y = [l['loss'] for l in chat_pretrain_loss['loss']]
```

```python
line_width = 2

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=list(range(len(chat_pretrain_loss_y))),
    y=chat_pretrain_loss_y,
    line_width=line_width,
    line_color='rgb(88,164,195)',
    name='Phyto-Chatbot',
))

fig.update_layout(
    xaxis={'showline': True, 'linewidth': line_width, 'linecolor': 'rgb(0,0,0)', 'mirror': True, 'ticks': 'outside', 'tickwidth': line_width, 'range': [0, 1060]},
    yaxis={'showline': True, 'linewidth': line_width, 'linecolor': 'rgb(0,0,0)', 'mirror': True, 'ticks': 'outside', 'tickwidth': line_width, 'range': [1.75, 1.9]},
    plot_bgcolor='white',
    font_family='Arial',
    font_color='rgb(0,0,0)',
    font_size=20,
    width=1920,
    height=1080/2)
fig.update_traces(mode='lines+markers')

fig.show()
if SAVE_FIGS:
    file_prefix = f'supplementary_fig.1a.chatbot_pretrain_loss.line'.lower().replace(' ', '_')
    fig.write_image(f"output/{file_prefix}.pdf")
    fig.write_image(f"output/{file_prefix}.png")
```
