# Extended Data Fig. 6d–e and 7d–e: Analyst Agent evaluation

## Purpose

Display model comparisons (6d–e) and post-training/Knowledge Agent ablations (7d–e) across five species (d) and ten analysis scenarios (e).

## Inputs and provenance

The eight author-supplied inline dictionaries hold four component scores on a 0–25 scale. Multiplication by four displays each component on a 0–100 scale; it does not compute a new total or uncertainty interval. The comparison panels select nine systems and the ablation panels select eight configurations from the same 15-system tables. Original values, species order, colors, and annotations are preserved. These aggregate tables are not regenerated from the Fig. 2 primary-run inputs.

## Run and outputs

Run all cells with the project Python kernel. Set `PHYTOMNI_SAVE=1` to write `output/extended_data_fig.6d.pdf`, `6e.pdf`, `7d.pdf`, and `7e.pdf` (each basename has the `extended_data_fig.` prefix). Species panels are 8 × 12 inches; scenario panels are 30 × 10 inches. The default run displays plots without writing figures.

```python
metric_order = ['plan_score', 'tool_score', 'param_score', 'rate_score']
metric_label_map = {
    'plan_score': 'Plan',
    'tool_score': 'Tool',
    'param_score': 'Param',
    'rate_score': 'Rate'
}
species_order = [
    'Oryza sativa',
    'Triticum aestivum',
    'Zea mays',
    'Glycine max',
    'Arabidopsis thaliana'
]


combined_data = {}
for metric in metric_order:
    pivot = metrics[metric]
    for species in species_order:
        combined_data[(metric_label_map[metric], species)] = pivot[species]

combined_df = pd.DataFrame(combined_data)
combined_df.columns = pd.MultiIndex.from_tuples(
    combined_df.columns,
    names=['Score', 'Scenario']
)


rotated_df = combined_df.T
fig, ax = plt.subplots(figsize=(8, 12), dpi=300)
sns.heatmap(
    rotated_df,
    ax=ax,
    cmap=tropic_cmap,
    annot=True,
    fmt=".1f",
    annot_kws={"size": 15},
    linewidths=0,
    cbar=True,
    cbar_kws={"shrink": 1, "aspect": 30, "pad": 0.02}
)

ax.tick_params(axis="x", labelrotation=55, labelsize=11)
ax.tick_params(axis="y", labelsize=10)
ax.set_ylabel(None)
for i in range(1, 4):
    ax.axhline(y=i * 5, color='black', linewidth=1.5)


metric_ticks = [2, 7, 12, 17]
ax2 = ax.twinx()
ax2.set_ylim(ax.get_ylim())
ax2.set_yticks(metric_ticks)
ax2.set_yticklabels(['' for m in metric_order], fontsize=1)
ax2.tick_params(axis='y', length=0)


for label in ax.get_xticklabels():
    label.set_horizontalalignment("right")

plt.tight_layout()
if SAVE_FIGS:
    fig.savefig("output/extended_data_fig.6d.pdf")
plt.show()
```
