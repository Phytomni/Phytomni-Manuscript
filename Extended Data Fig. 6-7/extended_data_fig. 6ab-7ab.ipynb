# Extended Data Fig. 6a–b and 7a–b: Knowledge Agent evaluation

## Purpose

Compare nine model systems on gene-identification and text-similarity benchmarks (6a–b), and show 16 post-training/retrieval configurations in the ablation panels (7a–b).

## Inputs and provenance

The author-supplied inline tables are the figure inputs. Identification accuracy is stored as a proportion and displayed as a percentage for five species and the mixed-species set. The ten text-similarity metrics retain their original values; radar axes are independently min–max scaled within each panel and display their raw ranges. Repeated identical configuration rows in the supplied radar table are retained to preserve the original plotting calculation. No evaluator or model API is called.

## Run and outputs

Run all cells using the standard `ir` kernel with tidyverse, ggplot2, ggradar, and scales installed. Set `PHYTOMNI_SAVE=1` to write `output/extended_data_fig.6a.pdf`, `6b.pdf`, `7a.pdf`, and `7b.pdf` (each basename has the `extended_data_fig.` prefix). Bar panels are 16 × 9 inches; radar panels are 14 × 8 inches. The default run displays plots without writing figures.

```r
sp_order <- c("Oryza sativa", "Arabidopsis thaliana", "Zea mays",
              "Glycine max", "Triticum Aestivum", "Mix")


plot_bar_chart <- function(data, fig_flag, target_order, target_colors) {
  # Select the comparison or ablation configurations.
  df_sub <- data %>% filter(!!sym(fig_flag) == "yes")
  
  if(nrow(df_sub) == 0) {
    message(paste("No data matched for", fig_flag))
    return(NULL)
  }
  
  df_long <- df_sub %>%
    select(model, Arabidopsis.thaliana, Oryza.sativa, Zea.mays, Glycine.max, Triticum.Aestivum, Mix) %>%
    pivot_longer(
      cols = -model,
      names_to = "type",
      values_to = "TRUE_Ratio"
    ) %>%
    mutate(type = str_replace(type, "\\.", " ")) 
  
  current_models <- intersect(target_order, unique(df_long$model))
  
  df_long <- df_long %>%
    mutate(
      model = factor(model, levels = current_models),
      type  = factor(type, levels = sp_order)
    )
  
  p <- ggplot(df_long, aes(x = model, y = TRUE_Ratio, fill = model)) +
    geom_bar(stat = "identity", position = "dodge", color = "black", linewidth = 0.2) + 
    facet_wrap(~type, scales = "free_x", ncol = 2) +
    labs(x = NULL, y = "Accuracy (%)") +
    theme_classic(base_size = 14) +
    scale_y_continuous(limits = c(0, 1), expand = expansion(mult = c(0, 0.05)),
                       breaks = seq(0, 1, 0.25), 
                       labels = seq(0, 100, 25)) + 
    theme(
      axis.text.x = element_blank(),  
      axis.ticks.x = element_blank(), 
      axis.line.x = element_blank(),  
      panel.background = element_rect(fill = "white", color = NA),
      plot.background  = element_rect(fill = "white", color = NA),
      strip.text = element_text(size = 12, face = "italic", margin = margin(b = 5)), 
      strip.background = element_blank(), 
      legend.position = "right",      
      legend.title = element_blank()  
    ) +
    scale_fill_manual(values = target_colors[current_models])
  
  return(p)
}
```

```r


radar_metrics <- c("BLEU-1", "BLEU-2", "BLEU-3", "BLEU-4",
                   "ROUGE-1 F1", "ROUGE-2 F1", "ROUGE-L F1",
                   "BERTScore Precision", "BERTScore F1", "BERTScore Recall")

plot_radar_chart <- function(data, fig_flag, target_order, target_colors, suffix) {
  df_sub <- data %>% filter(!!sym(fig_flag) == "yes")
  
  if(nrow(df_sub) == 0) {
    message(paste("No data matched for", fig_flag))
    return(NULL)
  }
  
  plot_data <- df_sub %>%
    rename(group = model_name) %>%
    select(group, all_of(radar_metrics))
  
  range_labels <- plot_data %>%
    select(-group) %>%
    summarise(across(everything(), list(min = ~min(.), max = ~max(.)))) %>%
    pivot_longer(everything(), names_to = c("metric", ".value"), names_sep = "_") %>%
    mutate(label = paste0(metric, "\n(", round(min, 3), "-", round(max, 3), ")"))
  
  metric_names <- colnames(plot_data)[-1]
  new_names <- setNames(range_labels$label, range_labels$metric)
  colnames(plot_data)[-1] <- new_names[metric_names]
  
  plot_data_norm <- plot_data %>%
    mutate(across(-group, ~ rescale(.)))
  
  current_levels <- intersect(target_order, unique(plot_data_norm$group))
  plot_data_norm <- plot_data_norm %>%
    mutate(group = factor(group, levels = current_levels)) %>%
    arrange(group)
  
  group_levels <- levels(plot_data_norm$group)
  colors_sub <- target_colors[group_levels]
  
  p <- ggradar(
    plot_data_norm,
    group.line.width = 0.8,
    group.point.size = 1.5,
    grid.line.width = 0.4,
    grid.min = 0,
    grid.mid = 0.5,
    grid.max = 1,
    background.circle.colour = "white",
    gridline.min.colour = "gray80",
    gridline.mid.colour = "gray80",
    gridline.max.colour = "gray80",
    axis.label.size = 4,
    legend.text.size = 12,
    values.radar = c("", "", "")
  )
  
  p$coordinates$clip <- "off"
  
  p <- p +
    scale_color_manual(values = setNames(colors_sub, group_levels)) +
    scale_fill_manual(values = setNames(colors_sub, group_levels)) +
    theme(
      panel.background = element_rect(fill = "white", colour = NA),
      plot.background = element_rect(fill = "white", colour = NA),
      panel.border = element_blank(),
      axis.line = element_blank(),
      axis.text.y = element_blank(),
      axis.ticks.y = element_blank(),
      axis.title = element_blank(),
      legend.position = "right",
      legend.text = element_text(size = 12),
      legend.title = element_blank(),
      legend.key.size = unit(0.5, "cm"),
      legend.key.height = unit(0.5, "cm"),
      legend.key.width = unit(0.5, "cm"),
      legend.margin = margin(0, 0, 0, 0),
      legend.box.margin = margin(0, 0, 0, 0),
      legend.spacing.y = unit(0.2, "cm"),
      legend.direction = "vertical",
      plot.margin = margin(1, 3, 1, 3, "cm")
    )
  
  print(p)
  
  return(p)
}
```
